```python
import numpy as np
import pandas as pd

def cronbach_alpha(data):
    """
    Calculate Cronbach's alpha.
    Missing values are handled using complete cases.
    """
    data = data.dropna()

    k = data.shape[1]

    item_variances = data.var(axis=0, ddof=1)
    total_score = data.sum(axis=1)
    total_variance = total_score.var(ddof=1)

    alpha = (k / (k - 1)) * (
        1 - item_variances.sum() / total_variance
    )

    return alpha
```

```python
# ============================================================
# GENAI STUDENT SURVEY — MASTER ANALYSIS SETUP
# ============================================================

# -----------------------------
# 1. Import libraries
# -----------------------------
import pandas as pd
import numpy as np

from google.colab import auth
import gspread
from google.auth import default

from scipy.stats import pearsonr


# -----------------------------
# 2. Authenticate Google account
# -----------------------------
auth.authenticate_user()

creds, _ = default()
gc = gspread.authorize(creds)

print("Google authentication successful!")


# -----------------------------
# 3. Connect to Google Sheet
# -----------------------------
sheet_id = "1Ylm3K9ukmAXqzUqHg_dma3XKLe2qEHOQHHm9XyBdk2M"

sheet = gc.open_by_key(sheet_id)
worksheet = sheet.get_worksheet(0)

data = worksheet.get_all_records()

df = pd.DataFrame(data)

print("\nRAW DATA LOADED")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# -----------------------------
# 4. Create a clean copy
# -----------------------------
df_clean = df.copy()


# -----------------------------
# 5. Define Likert columns
#    Python positions 9 to 32
#    = 24 Likert items
# -----------------------------
likert_positions = range(9, 33)


# -----------------------------
# 6. Clean Likert responses
# -----------------------------
for i in likert_positions:

    df_clean.iloc[:, i] = (
        df_clean.iloc[:, i]
        .astype("string")
        .str.strip()
    )

    df_clean.iloc[:, i] = df_clean.iloc[:, i].replace({
        "agree": "Agree",
        "Agree": "Agree",

        "Strongly agree": "Strongly Agree",
        "Strongly Agree": "Strongly Agree",

        "disagree": "Disagree",
        "Disagree": "Disagree",

        "Strongly disagree": "Strongly Disagree",
        "Strongly Disagree": "Strongly Disagree",

        "neutral": "Neutral",
        "Neutral": "Neutral",
        "Neitral": "Neutral",

        "": pd.NA,
        "'': pd.NA
    })


print("\nLIKERT CLEANING COMPLETED")


# -----------------------------
# 7. Convert Likert to numeric
# -----------------------------
likert_mapping = {
    "Strongly Disagree": 1,
    "Disagree": 2,
    "Neutral": 3,
    "Agree": 4,
    "Strongly Agree": 5
}

df_numeric = df_clean.copy()

for i in likert_positions:
    df_numeric.iloc[:, i] = df_numeric.iloc[:, i].map(likert_mapping)


print("LIKERT NUMERIC CODING COMPLETED")


# -----------------------------
# 8. Define constructs
# -----------------------------

# ADL = AI Usage & Digital Literacy
ADL = df_numeric.iloc[:, 9:15]

# AE = Academic Experience
AE = df_numeric.iloc[:, 15:20]

# CE = Challenges & Ethical Issues
CE = df_numeric.iloc[:, 20:27]

# IS = Institutional Support
IS = df_numeric.iloc[:, 27:31]


# -----------------------------
# 9. Cronbach's Alpha function
# -----------------------------
def cronbach_alpha(data):

    # Complete cases for reliability calculation
    data = data.dropna()

    k = data.shape[1]

    item_variances = data.var(axis=0, ddof=1)

    total_score = data.sum(axis=1)

    total_variance = total_score.var(ddof=1)

    alpha = (k / (k - 1)) * (
        1 - item_variances.sum() / total_variance
    )

    return alpha


# -----------------------------
# 10. Reliability analysis
# -----------------------------
alpha_adl = cronbach_alpha(ADL)
alpha_ae = cronbach_alpha(AE)
alpha_ce = cronbach_alpha(CE)
alpha_is = cronbach_alpha(IS)

print("\nCRONBACH'S ALPHA")
print("----------------------------")
print("ADL:", round(alpha_adl, 3))
print("AE :", round(alpha_ae, 3))
print("CE :", round(alpha_ce, 3))
print("IS :", round(alpha_is, 3))


# -----------------------------
# 11. Create analysis dataset
# -----------------------------
df_analysis = df_numeric.copy()

df_analysis["ADL"] = ADL.mean(axis=1, skipna=True)
df_analysis["AE"] = AE.mean(axis=1, skipna=True)
df_analysis["CE"] = CE.mean(axis=1, skipna=True)
df_analysis["IS"] = IS.mean(axis=1, skipna=True)


# -----------------------------
# 12. Add outcome variables
# -----------------------------

# Column 31 = Q32
df_analysis["Overall_Perception"] = df_numeric.iloc[:, 31]

# Column 32 = Q33
df_analysis["Continuance_Intention"] = df_numeric.iloc[:, 32]


# -----------------------------
# 13. Missing composite scores
# -----------------------------
print("\nMISSING COMPOSITE SCORES")
print("----------------------------")
print(
    df_analysis[
        ["ADL", "AE", "CE", "IS"]
    ].isna().sum()
)


# -----------------------------
# 14. Construct descriptive statistics
# -----------------------------
constructs = [
    "ADL",
    "AE",
    "CE",
    "IS",
    "Overall_Perception",
    "Continuance_Intention"
]

construct_stats = (
    df_analysis[constructs]
    .astype(float) # Ensure all selected columns are numeric for describe()
    .describe()
    .T
)

construct_stats = construct_stats[
    ["count", "mean", "std", "min", "max"]
]

construct_stats.columns = [
    "N",
    "Mean",
    "SD",
    "Minimum",
    "Maximum"
]

print("\nCONSTRUCT DESCRIPTIVE STATISTICS")
print("--------------------------------")
display(construct_stats.round(3))


# -----------------------------
# 15. Pearson correlations
# -----------------------------
corr_variables = [
    "ADL",
    "AE",
    "CE",
    "IS",
    "Overall_Perception",
    "Continuance_Intention"
]

correlation_results = []

for i in range(len(corr_variables)):

    for j in range(i + 1, len(corr_variables)):

        x = corr_variables[i]
        y = corr_variables[j]

        temp = df_analysis[[x, y]].dropna()

        r, p = pearsonr(
            temp[x],
            temp[y]
        )

        correlation_results.append({
            "Variable 1": x,
            "Variable 2": y,
            "N": len(temp),
            "Pearson r": round(r, 3),
            "p-value": round(p, 4)
        })


correlation_results = pd.DataFrame(
    correlation_results
)

print("\nPEARSON CORRELATIONS")
print("----------------------------")
display(correlation_results)


# -----------------------------
# 16. Final confirmation
# -----------------------------
print("\n===================================")
print("MASTER WORKFLOW COMPLETED")
print("===================================")
print("Respondents:", len(df))
print("Likert items:", len(likert_positions))
print("Missing Likert responses:",
      df_clean.iloc[:, 9:33].isna().sum().sum())
```

```python
# ============================================================
# PUBLIC VS PRIVATE UNIVERSITY COMPARISON
# ============================================================

comparison_vars = [
    "ADL",
    "AE",
    "ERAI",
    "IAPI",
    "Overall_Perception",
    "Continuance_Intention"
]

# Keep only valid university categories
uni_data = df_revised[
    df_revised[univ_col].isin(
        ["Public university", "Private University"]
    )
].copy()

print("University group sizes:")
print(uni_data[univ_col].value_counts())
```

```python
from scipy.stats import ttest_ind

results = []

for var in comparison_vars:

    public = pd.to_numeric(
        uni_data.loc[
            uni_data[univ_col] == "Public university",
            var
        ],
        errors="coerce"
    ).dropna()

    private = pd.to_numeric(
        uni_data.loc[
            uni_data[univ_col] == "Private University",
            var
        ],
        errors="coerce"
    ).dropna()

    # Welch's t-test
    t_stat, p_value = ttest_ind(
        public,
        private,
        equal_var=False
    )

    results.append({
        "Variable": var,
        "Public N": len(public),
        "Public Mean": public.mean(),
        "Private N": len(private),
        "Private Mean": private.mean(),
        "Mean Difference": private.mean() - public.mean(),
        "t": t_stat,
        "p-value": p_value
    })

uni_comparison = pd.DataFrame(results)

uni_comparison[
    [
        "Variable",
        "Public N",
        "Public Mean",
        "Private N",
        "Private Mean",
        "Mean Difference",
        "t",
        "p-value"
    ]
].round(3)
```

```python
# ============================================================
# UNDERGRADUATE VS MASTER'S COMPARISON
# ============================================================

level_data = df_revised[
    df_revised[level_col].isin(
        ["Undergraduate", "Master's"]
    )
].copy()

print("Study-level group sizes:")
print(level_data[level_col].value_counts())

level_results = []

for var in comparison_vars:

    undergraduate = pd.to_numeric(
        level_data.loc[
            level_data[level_col] == "Undergraduate",
            var
        ],
        errors="coerce"
    ).dropna()

    masters = pd.to_numeric(
        level_data.loc[
            level_data[level_col] == "Master's",
            var
        ],
        errors="coerce"
    ).dropna()

    t_stat, p_value = ttest_ind(
        undergraduate,
        masters,
        equal_var=False
    )

    level_results.append({
        "Variable": var,
        "Undergraduate N": len(undergraduate),
        "Undergraduate Mean": undergraduate.mean(),
        "Master's N": len(masters),
        "Master's Mean": masters.mean(),
        "Mean Difference": masters.mean() - undergraduate.mean(),
        "t": t_stat,
        "p-value": p_value
    })

level_comparison = pd.DataFrame(level_results)

display(
    level_comparison[
        [
            "Variable",
            "Undergraduate N",
            "Undergraduate Mean",
            "Master's N",
            "Master's Mean",
            "Mean Difference",
            "t",
            "p-value"
        ]
    ].round(3)
)
```

```python
# ============================================================
# FEMALE VS MALE COMPARISON
# ============================================================

gender_data = df_revised[
    df_revised[gender_col].isin(
        ["Female", "Male"]
    )
].copy()

print("Gender group sizes:")
print(gender_data[gender_col].value_counts())

gender_results = []

for var in comparison_vars:

    female = pd.to_numeric(
        gender_data.loc[
            gender_data[gender_col] == "Female",
            var
        ],
        errors="coerce"
    ).dropna()

    male = pd.to_numeric(
        gender_data.loc[
            gender_data[gender_col] == "Male",
            var
        ],
        errors="coerce"
    ).dropna()

    t_stat, p_value = ttest_ind(
        female,
        male,
        equal_var=False
    )

    gender_results.append({
        "Variable": var,
        "Female N": len(female),
        "Female Mean": female.mean(),
        "Male N": len(male),
        "Male Mean": male.mean(),
        "Mean Difference": male.mean() - female.mean(),
        "t": t_stat,
        "p-value": p_value
    })

gender_comparison = pd.DataFrame(gender_results)

display(
    gender_comparison[
        [
            "Variable",
            "Female N",
            "Female Mean",
            "Male N",
            "Male Mean",
            "Mean Difference",
            "t",
            "p-value"
        ]
    ].round(3)
)
```
